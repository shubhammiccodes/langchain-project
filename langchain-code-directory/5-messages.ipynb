## Messages

messages are the fundametal unit of context for models in lanchain. They represent input and output of a model, carrying both the content and metadata needed to represent the state of conversation when interacting with a LLM.

messages are object that contain:

ROLE: id entifies the message type(e.g system, user, assistant, tool, etc)

CONTENT: the actual text of the message(like text, images, audio, documents, etc.)

METADATA: optional fields such as response information, messageIDs, and token usage



Langchain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
from zoneinfo import reset_tzpath

from dotenv import load_dotenv

from langchain.chat_models import init_chat_model
load_dotenv()


model = init_chat_model(model="groq:openai/gpt-oss-120b")
```

```python
model.invoke("please tell me what is the role of messages in langchain and how they are used in LLMs?")
```

## Text Prompt

Text Prompts are strings - ideal for straightforward generation tasks where you don't need to retain conversation history.

```python
model.invoke("What is langchain?")
```

Use text prompt when:

1.you have a single , standalone request

2.you don't need conversation history or context

3.you want minimal code complexity and overhead

## Message Prompt


Alternatively, you can pass in a list of messages to the model by providing a list of message objects

Message types:

1. system message: tell the model how to behave and provide context for the conversation
2. Human message: represents input from the user, typically a question or request
3. AI message: represents output from the model, typically a response to a user input including text conent , tool calls and metadata
4. Tool message: represents output from a tool

System Messages:

a system message is a special type of message that provides instructions to the model about how it should behave and what context to use when generating responses.you can use a stem message to set the tone, define the model's role and establish guide lines for responses.

```python
from langchain.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage("you are poetry expert"),
    HumanMessage("Write a poem on Diwali festival in words of AI engineer")
]

response = model.invoke(messages)
response.content
```

```python
system_msg = SystemMessage("you are a  helpful coding assistant")

messages = [
    system_msg,
    HumanMessage("How do i create REST API?")
]
response = model.invoke(messages)
response.content
```

## Detailed info to the LLM through system messages

```python
from langchain.messages import SystemMessage, HumanMessage

system_msg = SystemMessage("""
You are a senior Python developer with expertise in web frameworks. Always provide code examples and explain your reasoning. Be concise but thorough in your explanations.""")

messages = [
    system_msg,
    HumanMessage("How do I create a REST API in Python?")
]

response = model.invoke(messages)
response.content
```

```python
from langchain.messages import AIMessage, HumanMessage, SystemMessage

ai_msg = AIMessage("I'll be happy to help you with that question")

messages = [
    SystemMessage("You are a helpful assistant."),
    HumanMessage("Can you help me ?"),
    ai_msg,
    HumanMessage("Great! What is 2+2 ?")
]

response = model.invoke(messages)
response.content
```

## Tool Messages

```python
from langchain.messages import AIMessage, ToolMessage


# After a model made a tool call
# (Here, we assume the model called a tool named "get_weather" with the argument "New York")

ai_msg = AIMessage(content=[],
                   tool_calls=[{
                        "name": "get_weather",
                        "args": {"city": "New York"},
                       "id": "tool_call_123"
                    }]
                   )


# execute the otool call and create the result message

weather_result = "Sunny, 72°F"
tool_msg = ToolMessage(
    content=weather_result,
    tool_name="get_weather",
    tool_call_id="tool_call_123"
)

#continue conversation
messages = [
    HumanMessage("What's the weather like in New York?"),
    ai_msg,  #model's tool call
    tool_msg,  # tool execution result
    ]

response = model.invoke(messages)
response.content
```
