### Model Intergration with Google Gemini, Ollama,  and GROQ

```python
import os

import langchain_groq
from dotenv import load_dotenv

load_dotenv()

os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')
os.environ['GROQ_API_KEY'] = os.getenv('GROQ_API_KEY')
```

## Google Gemini Model Integration

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(
    model="google_genai:gemini-3.8-flash")

model
```

```python
### invoke the model
response = model.invoke(" What are the vowels in alphabets? ?")
```

```python
response
```

```python
response.content
```

### Direct Google Gemini model integration

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash"
)
```

```python
response = model.invoke("What are the vowels in alphabets?")
response.content
```

### GROQ Model Integration

```python
import os

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

from langchain.chat_models import init_chat_model

model = init_chat_model(model="groq:openai/gpt-oss-120b")
response = model.invoke("why do parrots talk?")
response
```

```python
response.content
```

### Direct GROQ model integration

```python

import os

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(
    model="openai/gpt-oss-120b"
)
response = model.invoke(" what is the temperature in Delhi today?")
response.content
```

### Ollama Model Integration

```python
from langchain.chat_models import init_chat_model
from config import MODEL, MODEL_PROVIDER

#create the model instance
model = init_chat_model(
    model= MODEL,
    model_provider= MODEL_PROVIDER
)

response = model.invoke("What is the capital of France?")
response
```

### Direct Ollama model integration

```python
from langchain_ollama import ChatOllama
model = ChatOllama(
    model= MODEL,
    model_provider= MODEL_PROVIDER
)
response = model.invoke("Who is the owner of NVIDIA?")
response
```
