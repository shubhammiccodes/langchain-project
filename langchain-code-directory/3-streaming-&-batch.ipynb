## Streaming

Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. You can use a loop to process each chunk in real-time.

```python
import os

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

from langchain.chat_models import init_chat_model

model = init_chat_model(model="groq:openai/gpt-oss-120b")
```

```python
model.invoke("Write a 200 line poem about the beauty of nature.")
```

```python
model.stream("Write a 200 line how ai in revoltionizing the world of technology and innovation.")
```

```python
for chunk in model.stream("Write a 200 line how ai in revoltionizing the world of technology and innovation."):
    print(chunk.text, end="|", flush=True)
```

## Batch

Batch processing allows you to send multiple prompts to the model in a single request, which can be more efficient than sending them one at a time. The model will return a list of responses corresponding to each prompt. This is particularly useful when you have a large number of prompts to process, as it reduces the overhead of multiple network requests.

Batching a collectio of independendt request to a model can significantly improve performance and reduce costs, as the prcessingcan be done in parallel. This is especially beneficial when dealing with large datasets or when you need to generate multiple outputs for different inputs.

```python
responses = model.batch([
    "Write a 200 line poem about the beauty of nature.",
    "Write a 200 line how ai in revoltionizing the world of technology and innovation."
    "what is the future of ai in the next 10 years?"
])

for response in responses:
    print(response.text)
```

```python
model.batch([
    "Write a 200 line poem about the beauty of nature.",
    "Write a 200 line how ai in revoltionizing the world of technology and innovation."
    "what is the future of ai in the next 10 years?"
],
    config={'max_concurrency':5, #limit to 5 parallel requests}
            'max_retries':3, #retry failed requests up to 3 times
            'timeout':10 #timeout for each request in seconds
            }
)
```
